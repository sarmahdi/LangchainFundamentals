```python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import PromptTemplate
import gradio as gr
# Model  : gpt-4o-mini openAi


#1 Prompt Template

prompt_template_str ="""
Explain like I ama Java developer the Machine learning concept of topic of  {ML_term_concept}

"""


def generate_explanation(given_topic):
    #2 LangChain Prompt Template
    prompt_template = PromptTemplate.from_template(prompt_template_str)

    #3 initialize model
    model  = init_chat_model("gpt-4o-mini", model_provider="openai")

    #4 format the prompt
    #given_topic= "RAG"
    prompt = prompt_template.format(ML_term_concept=given_topic)
    #5 calling the model
    response =  model.invoke(prompt)
    return response.content

demo = gr.Interface(
    fn=generate_explanation,
    inputs=[gr.Textbox(label="Ai Term or Topic", lines =1)],
    outputs= [gr.Textbox(label = " Explanation", lines = 5)],
    name="AI term Explainer : "
)

demo.launch()
```
